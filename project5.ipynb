```python
def fetch_text(raw_url):
  """
  Fetches text content from a given URL, with caching mechanism.

  Parameters
  ----------
  raw_url : str
      The URL of the text file to fetch.

  Returns
  -------
  str
      The fetched text content if successful, otherwise an empty string.

  Notes
  -----
  - The function caches the fetched text locally to avoid re-downloading.
  - Cached files are stored in 'cs_110_content/text_cache/' and named based on a SHA1 hash of the URL.
  - Prints a success or failure message to standard output.
  - Includes a 10-second timeout for the HTTP request.
  """
  import requests
  from pathlib import Path
  import hashlib

  CACHE_DIR = Path("cs_110_content/text_cache")
  CACHE_DIR.mkdir(parents=True, exist_ok=True)

  def _url_to_filename(url):
    url_hash = hashlib.sha1(url.encode("utf-8")).hexdigest()[:12]
    return CACHE_DIR / f"{url_hash}.txt"

  cache_path = _url_to_filename(raw_url)

  SUCCESS_MSG = "✅ Text fetched."
  FAILURE_MSG = "❌ Failed to fetch text."
  try:
    if not cache_path.exists():
      response = requests.get(raw_url, timeout=10)
      response.raise_for_status()
      text_data = response.text
      cache_path.write_text(text_data, encoding="utf-8")
    print(SUCCESS_MSG)
    return cache_path.read_text(encoding="utf-8")

  except Exception as e:
    print(FAILURE_MSG)
    print(f"Error: {e}")
    return ""
```

```python
def print_text_stats(text):
  """
  Analyzes the provided text and prints statistics including character count, line count, and word count.

  Parameters
  ----------
  text : str
      The input text (string) for which to calculate statistics.

  Returns
  -------
  None
      This function does not return any value; it prints the statistics directly to standard output.

  Notes
  -----
  - Character count includes all characters, including whitespace and special characters.
  - Line count is determined by splitting the text by newline characters.
  - Word count is calculated by splitting each line by whitespace and summing the length of the resulting word lists.
    Words are considered to be sequences of non-whitespace characters.
  """
  num_chars = len(text)

  lines = text.splitlines()
  num_lines = len(lines)

  num_words = 0
  for line in lines:
    words_in_line = line.split()
    num_words_in_line = len(words_in_line)
    num_words += num_words_in_line

  print(f"Number of characters: {num_chars}")
  print(f"Number of lines: {num_lines}")
  print(f"Number of words: {num_words}")

print_text_stats(pride_prejudice_text)
```

```python
def get_word_counts(text):
  """
  Calculates the frequency of each word in the provided text.

  Parameters
  ----------
  text : str
      The input text (string) from which to count words.

  Returns
  -------
  dict
      A dictionary where keys are lowercase words and values are their corresponding counts.

  Notes
  -----
  - Words are split by whitespace characters.
  - All words are converted to lowercase before counting to ensure case-insensitive counting.
  """
  word_counts = {}
  lines = text.splitlines()
  for line in lines:
    words = line.split()
    for word in words:
      word = word.lower()
      if word in word_counts:
        word_counts[word] += 1
      else:
        word_counts[word] = 1
  return word_counts

word_counts = get_word_counts(pride_prejudice_text)
print(word_counts)
```

```python
def word_tokenization_normalization(text):
    """
    Tokenizes and normalizes a given text by lowercasing, removing stopwords, punctuation, numbers,
    and words shorter than 3 characters, then lemmatizing the remaining words.

    Parameters
    ----------
    text : str
        The input text (string) to be tokenized and normalized.

    Returns
    -------
    list
        A list of normalized (lowercase, lemmatized, filtered) words.

    Notes
    -----
    - Uses spaCy for tokenization, stop word removal, punctuation removal, and lemmatization.
    - Converts all words to lowercase.
    - Excludes tokens that are newlines, stopwords, punctuation, or numbers.
    - Excludes words with a length of 2 characters or less after stripping whitespace.
    """
    text = text.lower() # lowercase
    doc = nlp(text)     # loading text into model

    words_normalized = []
    for word in doc:
        if word.text != '\n' \
        and not word.is_stop \
        and not word.is_punct \
        and not word.like_num \
        and len(word.text.strip()) > 2:
            word_lemmatized = str(word.lemma_)
            words_normalized.append(word_lemmatized)

    return words_normalized
```

```python
def get_top_verbs(text):
  """
  Identifies and returns the 10 most frequent verbs in the given text.

  Parameters
  ----------
  text : str
      The input text (string) to analyze.

  Returns
  -------
  list
      A list of tuples, where each tuple contains a verb (lemma) and its frequency.

  Notes
  -----
  - Uses spaCy for part-of-speech tagging and lemmatization.
  - Converts all words to lowercase.
  - Filters tokens to include only verbs.
  """
  doc = nlp(text)
  verb_counts = {}
  for token in doc:
    if token.pos_ == 'VERB' and not token.is_stop and not token.is_punct and len(token.text.strip()) > 2:
      lemma = token.lemma_.lower()
      verb_counts[lemma] = verb_counts.get(lemma, 0) + 1

  sorted_verbs = sorted(verb_counts.items(), key=operator.itemgetter(1), reverse=True)
  return sorted_verbs[:10]
```

```python
top_verbs = get_top_verbs(pride_prejudice_text)
print("Top 10 most frequent verbs:")
for verb, count in top_verbs:
  print(f"{verb}: {count}")
```

```python
def word_count(word_list):
    """
    Calculates the frequency of each word in a given list of words.

    Parameters
    ----------
    word_list : list
        A list of words (strings) to count.

    Returns
    -------
    dict
        A dictionary where keys are lowercase words and values are their corresponding counts.

    Notes
    -----
    - All words are converted to lowercase before counting to ensure case-insensitive counting.
    """
    word_counts = {}
    for word in word_list:
      word = word.lower()
      if word in word_counts:
        word_counts[word] += 1
      else:
        word_counts[word] = 1
    return word_counts
```
